```python
import numpy as np
A=np.array([[1,-4],[-4,1]])
B=np.array([[2,-8],[-8,2]])
print("A+B",A+B)
print("A-(1/2)B=",A-(1/2)*B)
print("5A=",5*A)
```

```python
import  numpy as np
A=np.array([[2,3,1],[4,5,1]])
B=np.array([[-1,2],[4,-2],[3,6]])
print("AB=",np.dot(A,B))
print("BA=",np.dot(B,A))
```
